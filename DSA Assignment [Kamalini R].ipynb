1. Reverse an Array

```python
arr = [10, 20, 30, 40, 50]

i = len(arr) - 1

while i >= 0:
    print(arr[i], end=" ")
    i = i - 1
```

2. Count Even, Odd and Zero

```python
arr = [10, 5, 0, 7, 8, 0, 13, 4]

even = 0
odd = 0
zero = 0

for i in arr:
    if i == 0:
        zero = zero + 1
    elif i % 2 == 0:
        even = even + 1
    else:
        odd = odd + 1

print("Even:", even)
print("Odd:", odd)
print("Zero:", zero)
```

3. Sum of Positive and Negative Numbers

```python
arr = [10, -5, 20, -8, 15, -2]

positive = 0
negative = 0

for i in arr:
    if i > 0:
        positive = positive + i
    else:
        negative = negative + i

print("Positive Sum:", positive)
print("Negative Sum:", negative)
```

4. Remove Duplicate Elements

```python
arr = [10, 20, 10, 30, 20, 40, 30]

new = []

for i in arr:
    found = 0

    for j in new:
        if i == j:
            found = 1

    if found == 0:
        new.append(i)

print(new)
```

5. Find the Missing Number

```python
arr = [1, 2, 3, 5, 6, 7]

n = 7
total = 0

for i in range(1, n + 1):
    total = total + i

for i in arr:
    total = total - i

print("Missing Number:", total)
```

6. Rotate an Array

```python
arr = [1, 2, 3, 4, 5]
k = 2

n = len(arr)

for x in range(k):
    last = arr[n - 1]

    for i in range(n - 1, 0, -1):
        arr[i] = arr[i - 1]

    arr[0] = last

print(arr)
```

7. Find the Most Frequent Element

```python
arr = [2, 5, 2, 8, 5, 2, 3, 5, 2]

max_count = 0
most = 0

for i in arr:
    count = 0

    for j in arr:
        if i == j:
            count = count + 1

    if count > max_count:
        max_count = count
        most = i

print("Most Frequent Element:", most)
print("Frequency:", max_count)
```

8. Maximum Subarray Sum

```python
arr = [-2, 1, -3, 4, -1, 2, 1, -5, 4]

max_sum = arr[0]
current = arr[0]

start = 0
end = 0
temp = 0

for i in range(1, len(arr)):

    if current + arr[i] < arr[i]:
        current = arr[i]
        temp = i
    else:
        current = current + arr[i]

    if current > max_sum:
        max_sum = current
        start = temp
        end = i

print("Maximum Subarray Sum:", max_sum)

print("Subarray:", end=" ")
for i in range(start, end + 1):
    print(arr[i], end=" ")
```

9. Check Whether Two Arrays are Equal

```python
arr1 = [1, 2, 2, 3, 4]
arr2 = [4, 2, 1, 2, 3]

if len(arr1) != len(arr2):
    print("Arrays are Not Equal")
else:
    equal = 1

    for i in arr1:
        count1 = 0
        count2 = 0

        for j in arr1:
            if i == j:
                count1 = count1 + 1

        for j in arr2:
            if i == j:
                count2 = count2 + 1

        if count1 != count2:
            equal = 0

    if equal == 1:
        print("Arrays are Equal")
    else:
        print("Arrays are Not Equal")
```
